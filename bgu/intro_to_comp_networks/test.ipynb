```python
import networkx as nx
import numpy as np
```

```python
N = 10 ; p = 0.2 ; seed=None
G = nx.erdos_renyi_graph(N,p,seed)
for e in G.edges:
    G.edges[e]["weight"] = np.random.randint(0,10)
```

```python
pos = nx.circular_layout(G)
nx.draw(G,pos=pos,nodelist=range(N),with_labels=True)
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels,
    font_color='red' 
)
```

```python
dist_dict = nx.floyd_warshall(G)
dist_dict
# dist_dict[1]
```
